```python
query = """
SELECT Car_Name, Year, Selling_Price
FROM cars
WHERE Selling_Price > (
    SELECT AVG(Selling_Price)
    FROM cars
);
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
WITH AveragePrice AS (
    SELECT AVG(Selling_Price) AS Avg_Price
    FROM cars
)
SELECT Car_Name, Year, Selling_Price
FROM cars, AveragePrice
WHERE Selling_Price > Avg_Price;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT
    Car_Name,
    Year,
    Selling_Price,
    ROW_NUMBER() OVER (
        ORDER BY Selling_Price DESC
    ) AS Price_Rank
FROM cars;
"""

result = pd.read_sql(query, conn)
result.head(10)
```

```python
query = """
SELECT
    Car_Name,
    Year,
    Selling_Price,
    RANK() OVER (
        ORDER BY Selling_Price DESC
    ) AS Price_Rank
FROM cars;
"""

result = pd.read_sql(query, conn)
result.head(10)
```

```python
query = """
WITH RankedCars AS (
    SELECT
        Car_Name,
        Year,
        Selling_Price,
        RANK() OVER (
            ORDER BY Selling_Price DESC
        ) AS Price_Rank
    FROM cars
)
SELECT *
FROM RankedCars
WHERE Price_Rank <= 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
CREATE VIEW IF NOT EXISTS expensive_cars AS
SELECT
    Car_Name,
    Year,
    Selling_Price,
    Fuel_Type,
    Transmission
FROM cars
WHERE Selling_Price > 500000;
"""

conn.execute(query)
conn.commit()
```

```python
query = """
SELECT *
FROM expensive_cars;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT COUNT(*) AS Total_Expensive_Cars
FROM expensive_cars;
"""

result = pd.read_sql(query, conn)
result
```
